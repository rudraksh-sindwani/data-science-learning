```python
# Stock Market Data Analysis and Visualisation
# Part 1: Creating and Loading Stock Market Data

import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Create a date range
dates = pd.date_range(
    start="2025-01-01",
    periods=200,
    freq="D"
)

# Generate sample stock prices
closing_prices = 100 + np.cumsum(
    np.random.normal(0, 2, 200)
)

# Create the stock market dataset
data = pd.DataFrame({
    "Date": dates,
    "Open": closing_prices + np.random.normal(0, 1, 200),
    "High": closing_prices + np.random.uniform(1, 5, 200),
    "Low": closing_prices - np.random.uniform(1, 5, 200),
    "Close": closing_prices,
    "Volume": np.random.randint(100000, 1000000, 200)
})

# Display the first five rows
data.head()
```

# Stock Market Data Analysis and Visualisation

## Part 1: Creating and Loading Stock Market Data

### Key Observation

The dataset contains daily stock market information including opening price, highest price, lowest price, closing price, and trading volume.

This data will be used to analyse stock price trends, daily changes, returns, moving averages, and volatility.

```python
# Part 2: Dataset Inspection and Cleaning

# Display the dataset shape
print("Dataset Shape:", data.shape)

# Display dataset information
print("\nDataset Information:")
data.info()

# Check for missing values
print("\nMissing Values:")
print(data.isnull().sum())

# Check for duplicate rows
print("\nDuplicate Rows:", data.duplicated().sum())

# Display summary statistics
print("\nSummary Statistics:")
display(data.describe())
```

## Part 2: Dataset Inspection and Cleaning

### Key Observation

The dataset was inspected to understand its structure and data quality.

Missing values and duplicate rows were checked to identify potential data quality issues. Summary statistics provide an overview of the numerical stock market variables, including price and trading volume.

```python
# Part 3: Basic Stock Market Statistics

# Calculate basic statistics for stock prices

print("Average Opening Price:", round(data["Open"].mean(), 2))
print("Average Closing Price:", round(data["Close"].mean(), 2))

print("\nHighest Stock Price:", round(data["High"].max(), 2))
print("Lowest Stock Price:", round(data["Low"].min(), 2))

print("\nAverage Trading Volume:", round(data["Volume"].mean(), 0))

# Display summary statistics for price columns
price_statistics = data[
    ["Open", "High", "Low", "Close"]
].describe()

display(price_statistics)
```

## Part 3: Basic Stock Market Statistics

### Key Observation

Basic statistical analysis provides an overview of the stock's price behaviour during the analysed period.

The highest and lowest prices show the overall price range, while the average opening and closing prices provide an indication of the stock's typical value. Trading volume analysis also provides insight into the level of market activity.

```python
# Part 4: Visualising Stock Price Trends

import matplotlib.pyplot as plt

# Create a line chart for the closing price

plt.figure(figsize=(12, 6))

plt.plot(
    data["Date"],
    data["Close"]
)

plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.title("Stock Closing Price Trend Over Time")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
# Compare opening and closing prices over time

plt.figure(figsize=(12, 6))

plt.plot(
    data["Date"],
    data["Open"],
    label="Opening Price"
)

plt.plot(
    data["Date"],
    data["Close"],
    label="Closing Price"
)

plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.title("Opening vs Closing Stock Prices")

plt.xticks(rotation=45)

plt.legend()

plt.tight_layout()

plt.show()
```

## Part 4: Stock Price Trend Visualisation

### Key Observation

The line charts provide a visual representation of how stock prices changed over the analysed period.

The closing price trend helps identify overall movements and fluctuations in the stock value, while comparing opening and closing prices provides additional insight into daily market behaviour.

```python
# Part 5: Analysing Daily Price Changes

# Calculate the daily price change
data["Daily_Price_Change"] = data["Close"].diff()

# Calculate the daily percentage change
data["Daily_Percentage_Change"] = (
    data["Close"].pct_change() * 100
)

# Display the first few rows
data[
    [
        "Date",
        "Close",
        "Daily_Price_Change",
        "Daily_Percentage_Change"
    ]
].head(10)
```

```python
# Visualise daily price changes

plt.figure(figsize=(12, 6))

plt.plot(
    data["Date"],
    data["Daily_Price_Change"]
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Date")
plt.ylabel("Daily Price Change")
plt.title("Daily Stock Price Changes")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Part 5: Daily Price Change Analysis

### Key Observation

Daily price changes show how much the stock's closing price increased or decreased compared with the previous trading day.

Positive values indicate an increase in the closing price, while negative values indicate a decrease. Analysing these daily changes helps identify periods of higher price fluctuation.

```python
# Part 6: Calculate Moving Averages

# Calculate 7-day moving average
data["MA_7"] = data["Close"].rolling(
    window=7
).mean()

# Calculate 30-day moving average
data["MA_30"] = data["Close"].rolling(
    window=30
).mean()

# Display the relevant columns
data[
    [
        "Date",
        "Close",
        "MA_7",
        "MA_30"
    ]
].head(35)
```

```python
# Visualise stock prices and moving averages

plt.figure(figsize=(12, 6))

plt.plot(
    data["Date"],
    data["Close"],
    label="Closing Price"
)

plt.plot(
    data["Date"],
    data["MA_7"],
    label="7-Day Moving Average"
)

plt.plot(
    data["Date"],
    data["MA_30"],
    label="30-Day Moving Average"
)

plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.title("Stock Price with Moving Averages")

plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Part 6: Moving Average Analysis

### Key Observation

Moving averages help identify the underlying trend in stock prices by smoothing short-term fluctuations.

The 7-day moving average responds more quickly to recent price changes, while the 30-day moving average provides a broader view of the overall stock price trend.

```python
# Part 7: Trading Volume Analysis

# Calculate average trading volume
average_volume = data["Volume"].mean()

# Find the day with the highest trading volume
highest_volume_day = data.loc[
    data["Volume"].idxmax()
]

print("Average Trading Volume:", round(average_volume, 0))

print("\nHighest Trading Volume:")
print("Date:", highest_volume_day["Date"])
print("Volume:", highest_volume_day["Volume"])
```

```python
# Visualise trading volume over time

plt.figure(figsize=(12, 6))

plt.bar(
    data["Date"],
    data["Volume"]
)

plt.xlabel("Date")
plt.ylabel("Trading Volume")
plt.title("Trading Volume Over Time")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

## Part 7: Trading Volume Analysis

### Key Observation

Trading volume provides an indication of the level of market activity over time.

The analysis identifies the average trading volume and the date with the highest trading activity. Periods with higher volume indicate increased market participation compared with periods of lower trading volume.

## Part 8 — Highest and Lowest Stock Prices

The objective of this section is to identify the days when the stock reached its highest and lowest prices.

This helps understand the overall price range and the points of maximum and minimum market value in the dataset.

```python
highest_price = data.loc[data["High"].idxmax()]
lowest_price = data.loc[data["Low"].idxmin()]

print("Highest Stock Price:")
print("Date:", highest_price["Date"])
print("Price:", round(highest_price["High"], 2))

print("\nLowest Stock Price:")
print("Date:", lowest_price["Date"])
print("Price:", round(lowest_price["Low"], 2))
```

```python
plt.figure(figsize=(12, 6))

plt.plot(data["Date"], data["High"], label="High Price")
plt.plot(data["Date"], data["Low"], label="Low Price")

plt.xlabel("Date")
plt.ylabel("Stock Price")
plt.title("Highest and Lowest Stock Prices Over Time")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Key Observation

The analysis identifies the dates with the highest and lowest stock prices.

The difference between these values shows the overall price range of the simulated stock during the analysis period. This provides a simple view of the stock's price volatility.

## Part 9 — Analyse Stock Price Volatility

The objective of this section is to measure how much the stock price changes from day to day.

Daily percentage changes are used to calculate the standard deviation of returns, which provides a simple measure of price volatility.

```python
volatility = data["Daily_Percentage_Change"].std()

print("Daily Price Volatility:", round(volatility, 2), "%")
```

```python
plt.figure(figsize=(12, 6))

plt.hist(data["Daily_Percentage_Change"].dropna(), bins=30)

plt.xlabel("Daily Percentage Change (%)")
plt.ylabel("Frequency")
plt.title("Distribution of Daily Stock Returns")

plt.tight_layout()
plt.show()
```

### Key Observation

The standard deviation of daily percentage changes provides an estimate of the stock's daily volatility.

A higher volatility value indicates larger day-to-day price movements, while a lower value indicates more stable price movements.

## Part 10 — Analyse Monthly Stock Performance

The objective of this section is to group the stock data by month and analyse the average closing price.

Monthly aggregation helps identify broader price trends that may not be obvious from daily movements.


```python
data["Month"] = data["Date"].dt.to_period("M")

monthly_performance = data.groupby("Month")["Close"].mean()

display(monthly_performance)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(monthly_performance.index.astype(str), monthly_performance.values, marker="o")

plt.xlabel("Month")
plt.ylabel("Average Closing Price")
plt.title("Monthly Average Stock Closing Price")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Key Observation

The monthly analysis shows how the average closing price changes over the analysis period.

Aggregating daily prices into monthly averages makes it easier to identify longer-term upward and downward trends in the simulated stock data.

## Part 11 — Analyse Trading Volume vs Price

The objective of this section is to investigate whether trading volume changes alongside stock prices.

Comparing trading volume with closing prices can help identify periods of increased market activity.

```python
volume_price_correlation = data["Volume"].corr(data["Close"])

print("Correlation between Trading Volume and Closing Price:",
      round(volume_price_correlation, 3))
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(data["Volume"], data["Close"])

plt.xlabel("Trading Volume")
plt.ylabel("Closing Price")
plt.title("Trading Volume vs Closing Price")

plt.tight_layout()
plt.show()
```

### Key Observation

The correlation measures the linear relationship between trading volume and closing price.

A value close to 1 indicates a strong positive relationship, a value close to -1 indicates a strong negative relationship, and a value close to 0 indicates a weak linear relationship.

## Part 12 — Calculate Cumulative Returns

The objective of this section is to calculate the cumulative return of the stock over time.

Cumulative returns show how the investment value would have changed if the stock were held throughout the analysis period.

```python
data["Cumulative_Return"] = (
    (1 + data["Daily_Percentage_Change"] / 100)
    .cumprod() - 1
) * 100

data[["Date", "Close", "Daily_Percentage_Change", "Cumulative_Return"]].head(10)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(data["Date"], data["Cumulative_Return"])

plt.axhline(y=0, linestyle="--")
plt.xlabel("Date")
plt.ylabel("Cumulative Return (%)")
plt.title("Cumulative Stock Return Over Time")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
final_return = data["Cumulative_Return"].iloc[-1]

print("Final Cumulative Return:",
      round(final_return, 2), "%")
```

### Key Observation

The cumulative return shows the overall percentage change in the simulated investment over the analysis period.

Positive values indicate an overall gain compared with the starting price, while negative values indicate an overall loss.

## Part 13 — Identify Best and Worst Trading Days

The objective of this section is to identify the days with the largest positive and negative percentage price movements.

This helps highlight the strongest gains and largest losses during the analysis period.

```python
best_day = data.loc[data["Daily_Percentage_Change"].idxmax()]
worst_day = data.loc[data["Daily_Percentage_Change"].idxmin()]

print("Best Trading Day:")
print("Date:", best_day["Date"])
print("Closing Price:", round(best_day["Close"], 2))
print("Daily Change:", round(best_day["Daily_Percentage_Change"], 2), "%")

print("\nWorst Trading Day:")
print("Date:", worst_day["Date"])
print("Closing Price:", round(worst_day["Close"], 2))
print("Daily Change:", round(worst_day["Daily_Percentage_Change"], 2), "%")
```

```python
best_worst = data.nlargest(5, "Daily_Percentage_Change")

plt.figure(figsize=(10, 6))

plt.bar(
    best_worst["Date"].dt.strftime("%Y-%m-%d"),
    best_worst["Daily_Percentage_Change"]
)

plt.xlabel("Date")
plt.ylabel("Daily Percentage Change (%)")
plt.title("Largest Positive Daily Stock Movements")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Key Observation

The analysis identifies the best and worst trading days based on daily percentage changes.

These extreme movements provide useful information about periods of unusually strong gains or losses in the simulated stock data.

This section summarises the main findings from the stock market analysis.

The analysis examined price trends, daily price changes, moving averages, trading volume, volatility, monthly performance, and cumulative returns.
