```python
import matplotlib.pyplot as plt
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.bar(years,runs)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Sachin Tendulkar")
```

```python
import numpy as np
 
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x= np.arange(len(years))  # index positions
width=0.25

plt.bar(x+width, sachin, width=width, label="Sachin")
plt.bar(x, sehwag, width=width, label="Sehwag")
plt.bar(x-width, kohli, width=width, label="Kohli")

plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x,years)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np

years=[1995,2000,2005,2010,2015]
sachin=[1000,1200,1400,1500,1100]
kohli=[0,500,700,1300,1500]
sehwag=[300,500,1000,1200,0]

x= np.arange(len(years)) 
width=0.25

plt.bar(x-width, sachin,width=width, color="skyblue",label="Sachin")
plt.bar(x, kohli,width=width, color="red",label="Kohli")
plt.bar(x+width, sehwag,width=width, color="green", label="Sehwag")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Run comparision over 5 years")
plt.xticks(x,years)
plt.legend()
plt.tight_layout()
```

```python
players=["Sachin","Kohli","Sehwag"]
totalruns=[sum(sachin), sum(kohli), sum(sehwag)]

plt.barh(players, totalruns,color=["skyblue","red","green"])
plt.xlabel("Total runs")
plt.title("Total runs in debut 5 years")
plt.tight_layout()
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [500+700+1100+1500+1800, 0+200+900+1400+1600, 0+0+500+800+1100, 300+600+800+1100+900]

plt.barh(players, runs_5yrs, color="skyblue")
plt.xlabel("Total runs in 5 years")
plt.title("Run comparision over 5 years")
plt.tight_layout()
```

```python
players=["Sachin", "Sehwag", "Kohli"]
runs=[1000,2000,1200]

plt.bar(players,runs, color="skyblue")

for i in range(len(players)):
    plt.text(i, runs[i]+30, str(runs[i]), ha="center") 

plt.ylabel("RUNS")
plt.title("Runs scored by players")
plt.tight_layout()
```
