```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
num_rows=1000

data= {
    'camera':np.random.randint(8,201,size=num_rows),
    'age':np.random.randint(0,4,size=num_rows),
    'ram':np.random.choice([4,6,8,12,16],size=num_rows),
    'cpu':np.random.randint(40,101,size=num_rows),
    'sdcard':np.random.randint(0,2,size=num_rows),
    'sims':np.random.choice([1,2],size=num_rows),
    'prices':np.random.randint(8000,70001,size=num_rows),}

df = pd.DataFrame(data)

df.to_csv("smartphoneData.csv",index=False)
```
