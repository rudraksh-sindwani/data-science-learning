# 1. Understanding the Classification Problem

## Classification Problem

The aim of this task is to predict whether a small molecule is toxic or non-toxic to honey bees. The dataset used is ApisTox. It contains information about pesticides and other agrochemicals.

## Input Variables

The dataset contains information such as the source of the data, year, toxicity type, herbicide, fungicide, insecticide and other agrochemical information. Some columns such as name, CID, CAS and SMILES are identification or chemical description fields and will not be used as normal model features.

## Output Variable

The output variable is 'label'. It has two classes. A value of 0 means non-toxic and a value of 1 means toxic. Therefore this is a binary classification problem.

## Practical Significance

This classification task is useful because pesticide toxicity is important for protecting honey bees and supporting environmental and agricultural research. The UCI Machine Learning describes ApisTox as a dataset for pesticide toxicity to honey bees and lists classification as its associated task. 

## Consequences of Classification Errors

A wrong prediction can have important consequences. If a toxic chemical is predicted as non-toxic it could be treated as safe when it may harm honey bees. If a non-toxic chemical is predicted as toxic it could also lead to an incorrect decision about the chemical.

## Binary or Multiclass Classification

This is a binary classification problem because the target has two classes. The dataset contains 739 records in class 0 and 296 records in class 1. The classes are not balanced. Because of this imbalance accuracy alone may not give a complete view of model performance. Precision, recall, F1score and balanced accuracy should also be used.

## Evidence

The dataset contains 1035 records and 13 variables. The target has 739 class 0 records and 296 class 1 records. The official ApisTox description confirms that the binary label represents non-toxic and toxic compounds and that the dataset is designed for classification.

# 2. Data Exploration and Preparation

## Data Quality Assessment

The dataset contains 1035 rows and 13 columns. It has both categorical and numerical variables. No missing values were found in any column.

The target variable has 739 records in class 0 and 296 records in class 1. This shows that the dataset is not balanced.

## Data Pre-processing

The columns `name`, `CID`, `CAS` and `SMILES` were removed because they are mainly used to identify or describe the chemical compounds. The `ppdb_level` column was also removed because it has a very strong relationship with the target and may cause target leakage.

The categorical variables `source` and `toxicity_type` were converted into numerical variables using one-hot encoding. This method was selected because the categories do not have a natural order.

The data was divided into training and testing sets using a 70:30 ratio. Stratified splitting was used to keep a similar class distribution in both sets.

Standard scaling was applied to the features. The scaler was fitted using only the training data and then applied to the test data.

## Exploratory Analysis

The average feature values were compared between the two classes. The `insecticide` feature showed the largest difference between the classes. Its average value was 0.084 for class 0 and 0.557 for class 1. The `herbicide` feature also showed a clear difference.

The correlation results showed that `insecticide` had the strongest relationship with the target with a correlation of 0.52. `herbicide` had a correlation of -0.33 with the target.

The largest correlation between two input features was about -0.37. This shows that some features are related but there was no very strong relationship between the input features.

## Pre-processing Limitations

The dataset has no missing values so no imputation method was required. The removal of identifier columns means that some chemical information is not used by the models. The `ppdb_level` column was also removed to reduce the risk of target leakage. These decisions may reduce some available information but they make the model evaluation more reliable.

# 3. Model Development and Selection

## Logistic Regression

Logistic Regression was selected as the first model because it is a simple model for binary classification. It is easy to understand and can be used to predict the two classes in the dataset.

Five-fold stratified cross-validation was used to check the model. The mean F1 score was 0.638. Different values of C were tested. The best mean F1 score was 0.642 with C = 10. C controls the strength of the model and can affect model complexity.

The final Logistic Regression model had a training F1 score of 0.635 and a testing F1 score of 0.627. The two scores were close. This suggests that the model generalised well and did not show a strong sign of overfitting.

## Support Vector Machine

Support Vector Machine was selected as the second model because it is a useful classification method and was studied in Week 7. SVM finds a boundary between classes and can work well with scaled data.

Fivefold stratified cross-validation was used. The mean F1 score of the basic SVM was 0.653. Different C values were tested and the best mean F1 score was 0.675 with C=10.

The final SVM model had a training F1 score of 0.675 and a testing F1 score of 0.667. The scores were close which suggests that the model generalised well without a strong sign of overfitting.

## Model Comparison

The Logistic Regression model had a testing F1 score of 0.627. The SVM model had a testing F1 score of 0.667. Therefore the SVM performed better on the current 70:30 test split.

The SVM will be tested further after class imbalance is addressed. A final model should not be selected using only one result because class imbalance and the 80:20 split also need to be considered.

# 4. Addressing Data Imbalance

The dataset is imbalanced because class 0 contains 71.4% of the records while class 1 contains 28.6%. This imbalance can cause a model to focus more on the larger class and miss some cases from the smaller class.

The `class_weight="balanced"` method was selected because it gives more importance to the smaller class without creating new data. This was applied to both Logistic Regression and SVM.

For Logistic Regression the F1 score increased from 0.627 to 0.667. Recall increased from 0.584 to 0.708. This means the balanced model was better at finding class 1 cases. Accuracy changed only slightly from 0.801 to 0.797.

For SVM the results did not change after balancing. The accuracy stayed at 0.810 and the F1 score stayed at 0.667.

Therefore balancing improved the practical usefulness of Logistic Regression because it improved recall and F1 score for the smaller class. For SVM there was no improvement in the test results.

# 5. Final Model Evaluation and Recommendation

## Final Test Performance

The balanced Logistic Regression and balanced SVM models were tested using the untouched 70:30 test data. The Balanced Logistic Regression achieved an accuracy of 79.74%, precision of 63.00%, recall of 70.79% and F1 score of 66.67%. The Balanced SVM achieved an accuracy of 81.03%, precision of 67.05%, recall of 66.29% and F1 score of 66.67%.

The SVM had higher accuracy and precision while Logistic Regression had higher recall. Both models had the same F1 score.

## Error Analysis

The Balanced Logistic Regression confusion matrix showed 185 correct class 0 predictions and 63 correct class 1 predictions. It made 37 false positive errors and 26 false negative errors.

The Balanced SVM confusion matrix showed 193 correct class 0 predictions and 59 correct class 1 predictions. It made 29 false positive errors and 30 false negative errors.

The SVM made fewer false positive errors while Logistic Regression made fewer false negative errors.

## 80:20 Split

The split was changed from 70:30 to 80:20 and both models were tested again. Logistic Regression achieved an accuracy of 79.23% and an F1 score of 65.60%. SVM achieved an accuracy of 80.68% and an F1 score of 65.52%.

The SVM was still better in accuracy with the 80:20 split. The results were close to the 70:30 results which suggests that the models are reasonably stable.

## Recommendation

The Balanced SVM is recommended because it had the highest accuracy on both test splits and higher precision. Logistic Regression had slightly better recall which may be useful when finding as many class 1 cases as possible is more important.

SVM may take more time to train than Logistic Regression. Therefore the final choice should also consider the time available and the main purpose of the model. For this dataset the Balanced SVM gives a good balance between predictive performance and generalisation.
