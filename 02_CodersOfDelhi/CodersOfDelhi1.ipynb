# Coders of Delhi 

## Data Science Internship Project

### Project Overview

The goal is to analyze user data using **pure Python**, without using libraries such as Pandas or NumPy.

In this project, I will work with JSON data containing information about:

Users
Friend connections
Pages liked by users

---

## Project Workflow

This project is divided into four main stages:

### 1. Data Loading and Exploration
Load the JSON dataset using Python and explore the structure of users, connections, and liked pages.

### 2. Data Cleaning and Structuring
Clean the dataset by handling missing values, removing duplicate entries, removing inactive users, and standardizing the data.

### 3. People You May Know
Build a recommendation system that suggests potential friends based on mutual connections.

### 4. Pages You Might Like
Build a simple recommendation system that suggests pages based on similar user interests and interactions.

````python
#function to load the data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
````

````python
data= load_data("data.json")
````

````python
# write a function to display user and their connections
def display_users(data):
    print("user and their information")
    for user in data["users"]:
        print(f" Id: {user["id"]} {user["name"]} is friends with {user["friends"]} and liked pages are: {user["liked_pages"]}")
    print("\npages information")
    for page in data["pages"]:
        print(f" Id{page["id"]}: {page["name"]}") 

display_users(data)
````
