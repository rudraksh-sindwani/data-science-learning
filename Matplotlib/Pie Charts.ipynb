```python
import matplotlib.pyplot as plt
```

```python
plt.style.use("ggplot")
```

```python
labels=["Sachin", "Sehwag", "Kohli","Yuvraj"]
runs=[18000,10000,15000,11000]

plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels)
plt.show()
```

```python
colors=["red","skyblue","green","yellow"]
plt.pie(runs, labels=labels, colors=color)
plt.show()
```

```python
colors=["red","skyblue","green","yellow"]
plt.pie(runs, labels=labels, colors=color,
        wedgeprops={'edgecolor':"black",'linewidth':"2",'linestyle':"-."})
plt.show()
```

```python
explode=[0.1,0,0,0]

plt.pie(runs, labels=labels, colors=color, explode=explode, shadow=True,startangle=100,autopct="%1.2f%%")
plt.title("Exploded pie chart")
plt.show()
```
