# Breast Cancer Classification

## Part 1: Loading the Dataset

### Objective

The objective of this project is to analyse the Breast Cancer dataset and develop a machine learning model that can classify tumours as malignant or benign.

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Load the dataset
cancer = load_breast_cancer()

# Create a DataFrame
data = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

# Add the target variable
data["target"] = cancer.target

# Display the first five rows
data.head()
```

### Key Observation

The Breast Cancer dataset was successfully loaded into a Pandas DataFrame. It contains multiple numerical features describing characteristics of cell nuclei.

The target variable represents the diagnosis category and will be used to build a machine learning classification model.

## Part 2: Dataset Inspection and Data Quality Analysis

### Objective

This section examines the structure and quality of the Breast Cancer dataset. The analysis includes checking the dataset dimensions, data types, missing values, and summary statistics of the numerical features.

```python
# Display dataset dimensions
print("Dataset Shape:", data.shape)

# Display dataset information
print("\nDataset Information:")
data.info()

# Check for missing values
print("\nMissing Values:")
print(data.isnull().sum())

# Display summary statistics
print("\nSummary Statistics:")
display(data.describe())
```

### Key Observation

The Breast Cancer dataset contains multiple numerical features describing characteristics of cell nuclei, along with a target variable representing the tumour diagnosis.

The dataset contains no missing values, making it suitable for machine learning analysis without requiring additional missing-value imputation. The summary statistics also provide an overview of the different ranges and distributions of the numerical features.

## Part 3: Exploratory Data Visualisation

### Objective

This section uses visualisation techniques to explore the distribution of selected features in the Breast Cancer dataset and identify patterns between malignant and benign tumours.

```python
import matplotlib.pyplot as plt

# Plot histograms for selected features
selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area"
]

data[selected_features].hist(figsize=(10, 8))

plt.suptitle("Distribution of Selected Breast Cancer Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

for target in data["target"].unique():
    subset = data[data["target"] == target]
    
    plt.scatter(
        subset["mean radius"],
        subset["mean texture"],
        label=cancer.target_names[target]
    )

plt.xlabel("Mean Radius")
plt.ylabel("Mean Texture")
plt.title("Mean Radius vs Mean Texture")
plt.legend()

plt.show()
```

### Key Observation

The visualisations show that the numerical features have different distributions and ranges.

The scatter plot demonstrates that measurements such as mean radius and mean texture can provide useful information for distinguishing between malignant and benign tumours. However, some observations may overlap, indicating that multiple features may be needed for accurate classification.
